# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Both graphs are insightful as they provide different views of the same data. The first graph is useful in gauging visitor frequency trends in chronological order. There is a clear split in the weeks which allows for segmentation of the data but it still follows a continuous order so it is easy to gain a idea of visitor activity over time. The disadvantage is that the differing y-axes may be misleading. The second graph is useful in directly comparing each week's visitor frequency against each other. Plotting together on the same x-axis allows for easy comparison between the two lines, with the color differentiation supporting this as well. A disadvantage is that its hard to measure visitor activity over consecutive weeks with this view. An initial observation that is easily visible in this second graph is that both weeks saw their peak visitation in the middle of the week, between days 2-4, which may be of interest to the museum. Likewise, an initial observation of the first graph is that the distribution shape of the visitors between weeks seem to follow a cyclical pattern, at least in these two weeks.

```python
def dollar_format(x, pos):
    return f'${x:,.0f}'
```

```python
import matplotlib.ticker as mticker
```

```python
#Figure 8.5, pp. 193

x = np.arange(2008, 2015)
a = [400, 425, 425, 400, 300, 275, 250]
b = [375, 410, 415, 390, 295, 270, 240]
c = [np.nan, np.nan, 100, 190, 200, 225, 190]
d = [np.nan, np.nan, np.nan, 150, 250, 225, 215]
e = [np.nan, np.nan, np.nan, np.nan, np.nan, 100, 210]

plt.plot(x, a, label = "Product A", color = "blue", linewidth = 2)
plt.plot(x, b, label = "Product B", color = "red", linewidth = 2)
plt.plot(x, c, label = "Product C", color = "green", linewidth = 2)
plt.plot(x, d, label = "Product D", color = "purple", linewidth = 2)
plt.plot(x, e, label = "Product E", color = "teal", linewidth = 2)

plt.xlabel("Year")
plt.ylabel("Price")
plt.gca().yaxis.set_major_formatter(mticker.FuncFormatter(dollar_format))
plt.ylim(0, 500)
plt.title("Average Retail Product Price per Year", fontweight = "bold")

plt.grid(axis='y')
plt.legend()
plt.show
```
